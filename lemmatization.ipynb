```python
# QA's, chatbtot, text summarization

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
words = ["eat", "eating", "eats", "ate", "eaten", "go", "going", "goes", "went", "gone"]
```

```python
for word in words:
    print(f"{word} : {lemmatizer.lemmatize(word, pos='v')}")
```

```python
# pov deafault set as 'n' for noun, but we can change it to verb by setting pos='v', also for adjective pos='a', adverb pos='r'

print(lemmatizer.lemmatize("going", pos='r'))
print(lemmatizer.lemmatize("going", pos='v'))
print(lemmatizer.lemmatize("going", pos='n'))
print(lemmatizer.lemmatize("going", pos='a'))
```
